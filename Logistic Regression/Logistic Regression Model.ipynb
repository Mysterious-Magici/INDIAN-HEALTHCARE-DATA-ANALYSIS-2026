```python
import pandas as pd
import numpy as np

df = pd.read_csv('fairness_auditing_healthcare_data.csv')


threshold = df['Fasting_Blood_Glucose__mg_dl'].median()
print(f"Using dynamically calculated threshold: {threshold}")

df['Glucose_Class'] = np.where(df['Fasting_Blood_Glucose__mg_dl'] > threshold, 1, 0)

X = df.drop(['Fasting_Blood_Glucose__mg_dl', 'Glucose_Class'], axis=1)
y = df['Glucose_Class']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

accuracy = accuracy_score(y_test, y_pred)
print(f"Accuracy Score: {accuracy * 100:.2f}%\n")

print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\n")
print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
import joblib

df = pd.read_csv('fairness_auditing_healthcare_data.csv')
df['Target_Diabetes'] = np.where(df['Fasting_Blood_Glucose__mg_dl'] > df['Fasting_Blood_Glucose__mg_dl'].median(), 1, 0)
df['Target_Cholesterol'] = np.where(df['Total_Cholesterol_mg_dL'] > df['Total_Cholesterol_mg_dL'].median(), 1, 0)
df['Target_Obesity'] = np.where(df['BMI (kg meter square)'] > df['BMI (kg meter square)'].median(), 1, 0)

df.to_csv('processed_multidisease_data.csv', index=False)
print("Data successfully saved as 'processed_multidisease_data.csv'.\n")

base_features = [
    'Age', 'Gender_Male', 'Socioeconomic_Status_Encoded', 
    'Insurance_Encoded', 'Hospital Type_Government', 'Hospital Type_Private'
]

disease_configs = {
    "Diabetes": {
        "target": "Target_Diabetes",
        "features": base_features + ['HbA1c_mmol_mol', 'Glycemic_Status_Encoded', 'BMI (kg meter square)', 
                                     'Symptom Frequent Urination', 'Symptom Increased Thirst', 'Symptom Blurred Vision']
    },
    "High Cholesterol": {
        "target": "Target_Cholesterol",
        "features": base_features + ['BMI (kg meter square)', 'Age', 'Symptom Fatigue', 'HbA1c_mmol_mol']
    },
    "Obesity Risk": {
        "target": "Target_Obesity",
        "features": base_features + ['BMI_Status_Encoded', 'Glycemic_Status_Encoded']
    }
}

for disease_name, config in disease_configs.items():
    print(f"========== Model: {disease_name} ==========")
    

    X = df[config['features']]
    y = df[config['target']]
    

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
    
  
    model = LogisticRegression(max_iter=1000, class_weight='balanced', random_state=42)
    

    model.fit(X_train, y_train)
    

    y_pred = model.predict(X_test)
    

    acc = accuracy_score(y_test, y_pred)
    print(f"Accuracy Score: {acc * 100:.2f}%")
    print("\nConfusion Matrix:")
    print(confusion_matrix(y_test, y_pred))
    print("\nClassification Report:")
    print(classification_report(y_test, y_pred))
    
    model_filename = f"{disease_name.replace(' ', '_').lower()}_lr_model.pkl"
    joblib.dump(model, model_filename)
    print(f"-> Model saved to {model_filename}\n")
```
