```python
import numpy as np
df["Age"] = df["Age"].replace(32, np.nan)
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
sns.histplot(df["Age"], kde=True, bins=20)
plt.xlabel("Age")
plt.ylabel("Count")
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style="whitegrid")
plt.figure(figsize=(6, 4))
ax = sns.histplot(data=df, x="HbA1c (mmol/mol)", kde=True)
for container in ax.containers:
    ax.bar_label(container)
plt.title("Distribution of HbA1c")
plt.xlabel("HbA1c (mmol/mol)")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
invalid = df["Total Cholesterol (mg/dL)"].isin([200, 240])
valid_values = df.loc[~invalid, "Total Cholesterol (mg/dL)"]
df.loc[invalid, "Total Cholesterol (mg/dL)"] = np.random.choice(
    valid_values,
    size=invalid.sum(),
    replace=True
)
plt.figure(figsize=(8, 5))
sns.histplot(
    df["Total Cholesterol (mg/dL)"],
    kde=True,
    bins=20
)
plt.xlabel("Total Cholesterol (mg/dL)")
plt.ylabel("Count")
plt.title("Distribution of Total Cholesterol")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    df["Total Cholesterol (mg/dL)"],
    bins=20,
    kde=True,
    color="royalblue",
    edgecolor="white",
    alpha=0.75
)
plt.title(
    "Distribution of Total Cholesterol",
    fontsize=16,
    fontweight="bold"
)
plt.xlabel(
    "Total Cholesterol (mg/dL)",
    fontsize=12,
    fontweight="bold"
)
plt.ylabel(
    "Count",
    fontsize=12,
    fontweight="bold"
)
plt.grid(axis="y", alpha=0.25)
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("NEW INDIA HEALTHCARE DATA 2026.csv")
hba1c_count = df["HbA1c mmol mol"].value_counts().sort_index()
print(hba1c_count)

# Plot distribution
hba1c_count.plot(kind="bar", figsize=(8, 5))

plt.title("Distribution of HbA1c")
plt.xlabel("HbA1c (mmol/mol)")
plt.ylabel("Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
